```python
import pandas as pd

# 1. Cargar el CSV
df = pd.read_csv('items.csv')

# 2. Crear la columna 'importe' usando eval()
df.eval('importe = precio * cantidad', inplace=True)

# 3. Filtrar usando query()
dfa = df.query("categoria == 'A'")

# 4. Calcular el promedio de importe por categoría en el df original
promedios = df.groupby('categoria')['importe'].mean()

print(promedios)
```

```python
import pandas as pd

# 1. Cargar archivos
dfc = pd.read_csv('clientes.csv')
dfo = pd.read_csv('ordenes.csv')

# 2. Realizar Left Merge
# Mantiene todos los clientes, aunque no tengan órdenes
df_merged = pd.merge(dfc, dfo, on='cliente_id', how='left')

# 3. Calcular monto total por cliente y limpiar NaNs
monto_por_cliente = df_merged.groupby('cliente_id')['monto'].sum().fillna(0)

print(df_merged)
print(monto_por_cliente)
```

```python
import pandas as pd

# 1. Leer el archivo
df = pd.read_csv('sensores.csv')

# 2. Estandarizar la columna 'ts' a formato datetime
df['ts'] = pd.to_datetime(df['ts'], dayfirst=True, format='mixed')

# 3. Llenar nulos en 'temperatura' con el promedio de la columna
df['temperatura'] = df['temperatura'].fillna(df['temperatura'].mean())

# 4. Eliminar filas con nulos en 'humedad'
df.dropna(subset=['humedad'], inplace=True)

# Mostrar el resultado
print(df)
```
