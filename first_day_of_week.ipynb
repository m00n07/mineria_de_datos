# What is the First Day of the Week?

Whether the Gregorian calendar shows Sunday or Monday as the first day of the week depends on where you live.

Do more countries start the week on Sunday or Monday? What about people? What about by continent?

The file `first-day-of-week.csv` shows the first day of the week for each territory. The file `population.csv` shows the population in the year 2020 for each territory in millions, and the file `four-regions.csv` specifies whether each territory is in asia, europe, africa, or the americas.

```python
import pandas as pd
df = pd.read_csv('first-day-of-week.csv')
df.head()
```

```python
pop = pd.read_csv('population.csv')
pop.head()
```

```python
regions = pd.read_csv('four-regions.csv')
regions.head()
```

### Project Ideas

- How many territories show Friday, Saturday, Sunday, and Monday as the `first_day` of the week?

- How many people start the week on Friday, Saturday, Sunday, and Monday?
	- Hint: This will involve a `merge`.

- Which of the `four_regions` predominantly start the week on Sunday? On Monday? Are there any regions that are more divided between Sunday and Monday?
	- Hint: This will also involve a `merge`.

```python
# How many territories show Friday, Saturday, Sunday, and Monday as the `first_day` of the week?
# Contamos cuántos territorios inician la semana cada día
conteo_territorios = df['first_day'].value_counts()

print("Cantidad de territorios por primer día de la semana:")
print(conteo_territorios)
```

```python
# How many people start the week on Friday, Saturday, Sunday, and Monday?
# 1. Unimos los datos de días con la población
df_pop_dia = df.merge(pop, on='alpha3', how='inner')

# 2. Sumamos la población (que está en millones) agrupando por el día
poblacion_por_dia = df_pop_dia.groupby('first_day')['population'].sum().sort_values(ascending=False)

print("Población total (en millones) según el inicio de la semana:")
display(poblacion_por_dia)
```

```python
# Which of the `four_regions` predominantly start the week on Sunday? On Monday? Are there any regions that are more divided between Sunday and Monday?
# 1. Unimos el dataframe previo con el de regiones
df_completo = df_pop_dia.merge(regions, on='alpha3', how='inner')

# 2. Creamos una tabla cruzada (Crosstab) para ver la distribución
distribucion_regional = pd.crosstab(df_completo['four_regions'], df_completo['first_day'])

print("Distribución del primer día por región (Número de territorios):")
display(distribucion_regional)

# Opcional: Verlo en porcentajes para identificar la predominancia
predominancia = distribucion_regional.div(distribucion_regional.sum(axis=1), axis=0) * 100
print("\nPorcentaje de predominancia por región:")
display(predominancia.round(2))
```
