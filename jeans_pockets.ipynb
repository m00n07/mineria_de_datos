# Jean pockets

Are women jean pockets ridiculously small? Let's look at the data!

The file `jean-pocket-measurements.csv` shows pocket measurements for 20 popular brands.
Four pairs of jeans from each brand were measured: men's and women's skinny and straight styles. All jeans were designated a 32-inch waistband.

The following measurements are included:
- `height_front`: maximum front pocket height (cm)
- `height_back`: maximum back pocket height (cm)
- `width_front`: maximum front pocket width (cm)
- `width_back`: maximum back pocket width (cm)

```python
import pandas as pd
df = pd.read_csv('jean-pocket-measurements.csv')
df.head()
```

### Project Ideas

- What is the average difference in pocket `height_front` between women's and men's jeans?

- Is there a significant difference in pocket `height_front` between skinny and straight styles within the same gender?

- How do back pocket sizes compare between women's and men's jeans?

- Measure your phone's height. What percentage of women's and men's jeans can comfortably fit your phone in the pockets?



```python
# What is the average difference in pocket `height_front` between women's and men's jeans?
# Agrupamos por género y calculamos el promedio de altura frontal
promedios_genero = df.groupby('gender')['height_front'].mean()

dif_altura = promedios_genero['men'] - promedios_genero['women']

print(f"Altura promedio frontal (Hombres): {promedios_genero['men']:.2f} cm")
print(f"Altura promedio frontal (Mujeres): {promedios_genero['women']:.2f} cm")
print(f"Diferencia: Los bolsillos de hombre son {dif_altura:.2f} cm más profundos en promedio.")
```

```python
# Is there a significant difference in pocket `height_front` between skinny and straight styles within the same gender?
# Agrupamos por género y estilo
estilos = df.groupby(['gender', 'style'])['height_front'].mean().unstack()

print("Comparación de altura frontal por estilo y género:")
display(estilos)

# Diferencia por estilo
dif_estilo_mujeres = estilos.loc['women', 'straight'] - estilos.loc['women', 'skinny']
print(f"En mujeres, los 'straight' son {dif_estilo_mujeres:.2f} cm más profundos que los 'skinny'.")
```

```python
# How do back pocket sizes compare between women's and men's jeans?
# Calculamos un área aproximada para el bolsillo trasero
df['area_back'] = df['height_back'] * df['width_back']

# Comparamos el área promedio por género
area_trasera = df.groupby('gender')['area_back'].mean()

print("Área promedio de bolsillos traseros (cm²):")
display(area_trasera)

radio_back = area_trasera['men'] / area_trasera['women']
print(f"Los bolsillos traseros de hombre son {radio_back:.2f} veces más grandes que los de mujer.")
```

```python
# Measure your phone's height. What percentage of women's and men's jeans can comfortably fit your phone in the pockets?
mi_telefono_alto = 14.7  # Puedes cambiar este valor según tu teléfono real

# Creamos una columna booleana que indica si cabe
df['cabe_telefono'] = df['height_front'] >= mi_telefono_alto

# Calculamos el porcentaje por género
fit_stats = df.groupby('gender')['cabe_telefono'].mean() * 100

print(f"Porcentaje de jeans donde cabe un teléfono de {mi_telefono_alto} cm:")
print(f"Hombres: {fit_stats['men']:.1f}%")
print(f"Mujeres: {fit_stats['women']:.1f}%")
```
